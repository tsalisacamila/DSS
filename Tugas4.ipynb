```python
import numpy as np

def step_function(net, theta):
    return 1 if net >= theta else 0

# Data OR (binary)
inputs = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
targets = np.array([0, 1, 1, 1])
# Hyperparameter
alpha = float(input("Masukkan nilai alpha (learning rate): "))
theta = float(input("Masukkan nilai theta (threshold): "))
# Inisialisasi bobot dan bias
weights = np.zeros(2)
bias = 0
# Training
epochs = 10
for epoch in range(epochs):
    for i in range(len(inputs)):
        net = np.dot(inputs[i], weights) + bias
        output = step_function(net, theta)
        error = targets[i] - output
        weights += alpha * error * inputs[i]
        bias += alpha * error
print(f"Bobot akhir: {weights}, Bias akhir: {bias}")
```

```python
import numpy as np
import pandas as pd

def step_function(net, theta):
    return 1 if net >= theta else 0

def encode_binary(row):
    outlook = {'Sunny': [1,0,0], 'Overcast': [0,1,0], 'Rain': [0,0,1]}[row['Outlook']]
    temp = {'Hot': [1,0,0], 'Mild': [0,1,0], 'Cool': [0,0,1]}[row['Temperature']]
    humidity = 1 if row['Humidity'] == 'High' else 0
    wind = 1 if row['Wind'] == 'Weak' else 0
    return outlook + temp + [humidity, wind]

def encode_bipolar(row):
    outlook = {'Sunny': [1,-1,-1], 'Overcast': [-1,1,-1], 'Rain': [-1,-1,1]}[row['Outlook']]
    temp = {'Hot': [1,-1,-1], 'Mild': [-1,1,-1], 'Cool': [-1,-1,1]}[row['Temperature']]
    humidity = 1 if row['Humidity'] == 'High' else -1
    wind = 1 if row['Wind'] == 'Weak' else -1
    return outlook + temp + [humidity, wind]

def encode_target(row, bipolar=False):
    if bipolar:
        return 1 if row['Play'] == 'Yes' else -1
    else:
        return 1 if row['Play'] == 'Yes' else 0

def train_perceptron(data, inputs, targets, alpha, theta, bipolar=False):
    weights = np.zeros(len(inputs[0]))
    bias = 0
    for epoch in range(20):
        for i in range(len(inputs)):
            net = np.dot(inputs[i], weights) + bias
            output = step_function(net, theta)
            error = targets[i] - output
            weights += alpha * error * inputs[i]
            bias += alpha * error
    return weights, bias

# Load dataset
df = pd.read_csv("https://raw.githubusercontent.com/tsalisacamila/DSS/465cef245d31b90a0ac0c68f51932e49611c7fb6/weather.csv")

alpha = float(input("Masukkan nilai alpha: "))
theta = float(input("Masukkan nilai theta: "))

# Binary encoding
X_binary = np.array([encode_binary(row) for index, row in df.iterrows()])
y_binary = np.array([encode_target(row) for index, row in df.iterrows()])
w_bin, b_bin = train_perceptron(df, X_binary, y_binary, alpha, theta)
print("=== Hasil Binary ===")
print(f"Bobot: {w_bin}, Bias: {b_bin}")
# Bipolar encoding
X_bipolar = np.array([encode_bipolar(row) for index, row in df.iterrows()])
y_bipolar = np.array([encode_target(row, bipolar=True) for index, row in df.iterrows()])
w_bip, b_bip = train_perceptron(df, X_bipolar, y_bipolar, alpha, theta)
print("=== Hasil Bipolar ===")
print(f"Bobot: {w_bip}, Bias: {b_bip}")
```
